```python
# 상대 경로(절대 경로는 자꾸 에러가 난다.)
bank_df = pd.read_csv('data/bank_20261001_20261031.csv', encoding='cp949')
bank_df.head()
```

# 1. 파일 읽기

```python
# glob은 파일명을 검색할 때 특수 기호(와일드카드)를 지원
# * (아무 글자나 허용)
# ? (한 글자 대체): 정확히 자릿수의 한 글자를 대체
import glob

baemin_files = sorted(glob.glob("data/baemin/baemin*.csv"))
baemin_list = []
for file in baemin_files:
    x = pd.read_csv(file)
    baemin_list.append(x)

baemin_df = pd.concat(baemin_list, ignore_index=True)
# baemin_df.info()

baemin_df
```

```python
coupang_files = sorted(glob.glob("data/coupang/coupang*.csv"))
coupang_list = []
for file in coupang_files:
    y = pd.read_csv(file)
    coupang_list.append(y)

coupang_df = pd.concat(coupang_list, ignore_index=True)
# coupang_df.info()

coupang_df
```

# 2. 칼럼명 통일

```python
baemin_df = baemin_df.rename(columns = {
    "order_at" : "주문일시" , 
    "total_bill" : "전체금액" , 
    "ad_yn" : "광고유입여부" ,
    "cpc": "광고집행비" ,
})

baemin_df["플랫폼"] = "배민"

baemin_df
```

```python
coupang_df = coupang_df.rename(columns = {
    "order_time" : "주문일시" , 
    "total_price" : "전체금액" , 
    "advertisement_yn" : "광고유입여부" , 
    "ad_pct" : "광고집행비" , 
})    

coupang_df["광고집행비"] = pd.to_numeric(coupang_df["광고집행비"], errors="coerce")
coupang_df["광고집행비"] = (coupang_df["전체금액"] * coupang_df["광고집행비"]) / 100
# coupang_df.isna().count()

coupang_df["플랫폼"] = "쿠팡"
coupang_df
```

# 3.데이터 통합

```python
total_df = pd.concat([baemin_df, coupang_df], ignore_index=True)
total_df["주문일시"] = pd.to_datetime(total_df["주문일시"])

total_df["주문일자"] = total_df["주문일시"].dt.date
total_df["주문일시"] = total_df["주문일시"].dt.strftime("%H:%M:%S")

total_df = total_df.sort_values(by=["주문일자", "주문일시"], ascending=True)
total_df = total_df.reset_index(drop=True)

total_df = total_df[["플랫폼", "주문일자",  "주문일시", "전체금액", "광고유입여부", "광고집행비"]]

total_df
```

# 4. 일자별 매출 집계

```python
daily_total = total_df.pivot_table(
    index="주문일자" , 
    columns="플랫폼" ,
    values ="전체금액" , 
    aggfunc="sum" , 
)
daily_total["합계"] = daily_total["배민"] + daily_total["쿠팡"]
daily_total
```

# 5. 광고 효과 분석

```python
mask = total_df["광고유입여부"] == "Y"
ad_total_df = total_df[mask]

total_count = total_df.groupby("플랫폼").size()
ad_total_count = ad_total_df.groupby("플랫폼").size()

rate = ad_total_count / total_count * 100
print(rate.round(1))

# ad_Spend : 광고비
ad_Spend = ad_total_df.groupby("플랫폼")["광고집행비"].sum()
print()
print(ad_Spend)
```

# 6. 식자재비 추출

```python
bank_df["거래일시"] = pd.to_datetime(bank_df["거래일시"])

bank_df["거래일자"] = bank_df["거래일시"].dt.date
bank_df["거래일시"] = bank_df["거래일시"].dt.strftime("%H:%M:%S")

bank_df = bank_df.sort_values(by=["거래일자", "거래일시"], ascending=True)
bank_df = bank_df[["거래일자", "거래일시", "적요", "거래처", "출금액", "입금액", "잔액"]]

bank_df
```

```python
# .copy 질문하기
mask = bank_df["거래처"] == "식자재마트"
food = bank_df[mask]

food["주문일자"] = food["거래일자"]
food.head()
```

```python
daily_food = food.groupby("주문일자")["출금액"].sum()
daily_food.name = "식자재비"
daily_food
```

# 7. 최종 결과표 저장

```python
daily_total
```

```python
result = daily_total.join(daily_food)

result["매출-식자재비"] = result["합계"] - result["식자재비"]

result = result.reset_index()

result

```

```python
result.to_csv("result.csv", index=False, encoding="utf-8-sig")
print("저장이 완료되었습니다.")
```
