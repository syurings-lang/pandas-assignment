### 5. 복합 조건

각각 조건을 만족하는 행 수를 출력하세요.

1. 3등석 **그리고** 남성 **그리고** 생존한 승객 (`&`)
2. 1등석 **또는** 요금 100 이상인 승객 (`|`)
3. 혼자 탑승하지 **않은** 승객 (`~`, `alone` 열 사용)

```python
# 1번 문제
result01 = (titanic["class"] == "Third") & (titanic["sex"] == "male") & (titanic["survived"] == 1)
print(result01.sum())
print("승객 수 :", len(titanic[result01]))

# 2번 문제
result02 = (titanic["class"] == "First") | (titanic["fare"] > 100)
print(result02.sum())
print("승객 수 :", len(titanic[result02]))

# 3번 문제
result03 = ~(titanic["alone"] == True)
print(result03.sum())
print("승객 수 :", len(titanic[result03]))
```

### 7. loc[조건, 열]

12살 이하 **그리고** 3등석 승객의 `age`, `sex`, `survived` 열만 `kids` 변수에 담고, 행 수와 생존률(`survived` 평균, 소수 둘째 자리)을 출력하세요.

```python
mask  = (titanic["age"] <= 12) & (titanic["class"] == "Third")
kids = titanic.loc[mask, ["age", "sex", "survived"]]
print("행 수 :", len(kids))

result_sur = kids["survived"].mean()
print("생존률 :", round(result_sur, 2))
```

### 10. 종합 — 20~40세 남성, 1·2등석 vs 3등석

질문: "20~40세 남성 승객의 생존률은 객실 등급에 따라 얼마나 다를까?"

1. 남성 · 나이 20~40세(`between`) · 1·2등석(`isin`) 조건으로 `pclass`, `age`, `fare`, `survived` 열을 선택하세요.
2. 행 수, 생존률(소수 둘째 자리), 요금 상위 3명을 출력하세요.
3. 같은 조건을 `query`로 써서 행 수를 검증하세요.
4. 등급만 3등석으로 바꿔 생존률을 구하고, 두 생존률을 비교하는 해석을 마크다운 셀에 2~3문장으로 쓰세요.

```python
# 1번 문제
mask_sex = titanic["sex"] == "male"
mask_age = titanic["age"].between(20, 40)
mask_pclass = titanic["pclass"].isin([1, 2])
result = titanic[mask_sex & mask_age & mask_pclass]
x = ["pclass", "age", "fare", "survived"]
print(result[x])

# 2번 문제 
print("행 수 :" , len(result))

mask_sur = titanic["survived"] == 1
result02 = titanic[mask_sex & mask_age & mask_pclass & mask_sur]
sur_len = len(result02["survived"])
survival_rate = (sur_len / 103) * 100
print("생존률 :", round(survival_rate, 2))


result_sort = result.sort_values(by="fare" , ascending=False)
print(result_sort.head(3))

# 3번 문제
result_q = titanic.query("sex == 'male' and age.between(20, 40) and pclass.isin([1, 2])")
print("행 수 :", len(result_q))

# 4번 문제
mask_pclass03 = titanic["pclass"].isin([3])
result03 = titanic[mask_sex & mask_age & mask_pclass03 & mask_sur]
sur_len03 = len(result03["survived"])
total = len(titanic[mask_sex & mask_age & mask_pclass03])
survival_rate03 = (sur_len03 / total) * 100
print("생존률 :", round(survival_rate03, 2))
```

### 20~40세 남성, 1,2등석의 생존률은 24.27% 이고 20~40세 남성 3등석의 생존률은 15.09 이다.3등석 생존률이 1,2등석 생존률보다 낮은 이유는 상등석은 배의 상부에 위치해 있어 탈출이 용이했고 하등석은 배의 하부에 있어서 사고 발생시 3등석의 사람들은 아래에서 올아와야 하는 구조적 불리함 때문데 생존률이 낮았을 것으로 예상된다.
